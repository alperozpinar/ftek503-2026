## [G] Girdiler

Excel'de girdiler Hesap sayfasının B1, B2 ve B3 hücrelerindedir. Python'da her girdiye bir **ad** verilir. Bu adlara **değişken** denir.

| Excel | Python | Değer |
|---|---|---|
| B1 (anapara) | `anapara` | 10000 |
| B2 (yıllık oran, %40) | `oran` | 0.40 |
| B3 (süre, yıl) | `yil` | 3 |

Okurken dikkat edilecek noktalar:
- `anapara = 10000` satırındaki `=` "eşittir" anlamına gelmez; sağdaki değeri soldaki ada verir. Buna **atama** denir. Excel'de formülün başındaki `=` ile karıştırılmamalıdır.
- `#` işaretinden sonrası **açıklamadır**. Python o kısmı çalıştırmaz; insanlar için yazılır.
- Python'da ondalık ayırıcı **noktadır**: `0.40`. Türkçe Excel'deki `0,40` Python'da `0.40` yazılır.
- %40, Python'da `0.40` yazılır. Excel'de de hücrede 0,40 saklanır; yüzde yalnız bir görünüştür.

```python
# [G] Girdiler: Excel'deki Hesap sayfasının B1, B2, B3 hücreleriyle aynı değerler
anapara = 10000   # TL          (Excel: B1)
oran = 0.40       # yıllık %40  (Excel: B2)
yil = 3           # yıl         (Excel: B3)
print(anapara, oran, yil)
```

## [H] Basit faiz

Basit faizde faiz yalnız anaparaya işler. Excel'de B5'e `=B1*(1+B2*B3)` yazılmıştır ve sonuç 22.000 TL'dir.

**Tahmin sorusu:** Aşağıdaki hücre ne yazdırır? Tahmin not edilir, sonra hücre çalıştırılır.

```python
# [H] Basit faiz. Excel (Türkçe ve İngilizce aynı): =B1*(1+B2*B3)
basit = anapara * (1 + oran * yil)
print(basit)
```

## [H] Bileşik faiz

Bileşik faizde her yılın faizi anaparaya eklenir ve ertesi yıl o da faiz kazanır. Excel'de B6'ya `=B1*(1+B2)^B3` yazılmıştır ve sonuç 27.440 TL'dir.

Python'da üs işareti `^` değildir; üs **iki yıldız** (`**`) ile yazılır.

**Tahmin sorusu:** Aşağıdaki hücre tam olarak `27440` mı yazdırır?

```python
# [H] Bileşik faiz. Excel: =B1*(1+B2)^B3 ; Python'da üs işareti ** (iki yıldız)
bilesik = anapara * (1 + oran) ** yil
print(bilesik)
```

### İnceleme: 27439.999999999993 sonucunun nedeni

- Bilgisayar ondalık sayıları ikilik sistemde saklar. 0.40 gibi bazı sayılar ikilik sistemde tam yazılamaz; çok küçük bir yuvarlama payıyla saklanır. Durum, 1/3'ün ondalık gösteriminin 0,3333... biçiminde bitmeden sürmesine benzer.
- Bu küçük pay çarpmalar boyunca taşınır. Sonuç 27.440'tan yaklaşık 0,000000000007 kadar sapar. Para hesabında bu sapma önemsizdir.
- Excel'de hücre iki ondalıkla biçimlendirildiği için orada 27.440,00 görünür.
- Bu nedenle Excel ile Python karşılaştırılırken "tam eşit mi?" sorusu değil, **"fark çok küçük mü?"** sorusu sorulur. Ayrıntısı H6'dadır.
- Python'da `^` üs anlamına gelmez, başka bir işlemdir. `(1 + oran) ^ yil` yazıldığında Python bir hata mesajı (`TypeError`) verir. Üs için her zaman `**` kullanılır.

## [S] Sağlama: Python ve Excel sonuçlarının karşılaştırılması

Üç yeni araç:
- `abs(x)`: x'in işaretini atar. Excel'deki MUTLAK (ABS) fonksiyonu gibidir. `abs(-3)` sonucu 3'tür.
- `==` (çift eşittir): iki tarafın eşit olup olmadığını **sorar**. Cevap `True` (doğru) ya da `False` (yanlış) olur; Excel'deki mantıksal değer gibidir (Hesap!B53). Tek `=` atamadır, çift `==` karşılaştırmadır.
- `assert koşul`: "bu koşul doğru olmalı" anlamındadır. Koşul doğruysa hiçbir şey olmaz. Yanlışsa Python `AssertionError` hatasıyla durur.

Aşağıda fark 0,01'den (1 kuruştan) küçükse sağlama geçer.

**Tahmin sorusu:** İlk `print` satırı `True` mu `False` mu yazdırır?

```python
# [S] Sağlama: Excel'in verdiği sonuçlarla karşılaştırma
excel_basit = 22000     # Excel Hesap!B5
excel_bilesik = 27440   # Excel Hesap!B6

print(bilesik == excel_bilesik)          # tam eşit mi?
print(abs(bilesik - excel_bilesik))      # aradaki fark ne kadar?
assert abs(basit - excel_basit) < 0.01       # fark 1 kuruştan küçükse sessizce geçer
assert abs(bilesik - excel_bilesik) < 0.01
print("Sağlama tamam: Python ile Excel aynı sonucu verir.")
```

**İnceleme:** İkinci satırdaki `7.275957614183426e-12` bilimsel gösterimdir: 7,28 × 10⁻¹², yani yaklaşık 0,00000000000728. Bu değer sıfır değildir, ancak 1 kuruştan çok küçüktür. Bu nedenle `==` `False` vermiş, `assert` ise geçmiştir.

## [H] Yıl yıl tablo

Excel'de B9'a `=B8*(1+$B$2)` yazılıp formül aşağı sürüklenmiştir. `$B$2` sayesinde oran her satırda aynı hücreden gelir. Python'da aynı işin her yıl için tekrarlanması `for` döngüsüyle yapılır.

**`for` döngüsü H6'da ele alınır.** Bu aşamada hücre yalnız çalıştırılır ve çıktı Excel'deki tabloyla (Hesap!A8:F11) karşılaştırılır.

`round(x, 2)`: x'i 2 ondalığa yuvarlar, Excel'deki YUVARLA (ROUND) gibi. İkisi arasındaki küçük fark H6'da ele alınır.

**Tahmin sorusu:** Son satırdaki bileşik bakiye kaç çıkar? Fark hangi yıldan itibaren sıfırdan büyük olur?

```python
# [H] Yıl yıl tablo. Excel: B9 =B8*(1+$B$2)   D9 =$B$1*(1+$B$2*A9)   F9 =B9-D9
# for döngüsü H6'da ele alınır; bu aşamada hücre yalnız çalıştırılır.
bakiye = anapara                          # Excel B8 =B1 (0. yıl)
print("Yıl", "Bileşik", "Basit", "Fark")
print(0, bakiye, anapara, 0)
for t in range(1, yil + 1):               # t sırayla 1, 2, 3 olur: Excel'de bir satır aşağı inmek gibi
    bakiye = bakiye * (1 + oran)          # oran her satırda aynı: Excel'deki $B$2
    basit_t = anapara * (1 + oran * t)    # basit faiz, t. yılın sonunda
    print(t, round(bakiye, 2), round(basit_t, 2), round(bakiye - basit_t, 2))
```

```python
# [S] Sağlama: tablonun son satırı tek hücre formülüne eşit mi? (Excel Hesap!B36: =B11-B6)
print(bakiye - bilesik)
assert abs(bakiye - bilesik) < 0.01
print("Tablonun son satırı tek hücre formülüyle aynı.")
```

## [H] TOPLA, ORTALAMA, MİN, MAK'ın Python karşılığı

Excel'de her yılın bileşik faizi Hesap!C9:C11 hücrelerindedir: 4.000, 5.600 ve 7.840 TL. Aşağıda önce her yılın bakiyesi Excel'deki gibi bir öncekinden hesaplanır (B9, B10, B11). Sonra her yılın faizi bulunur (C9:C11) ve köşeli parantez içinde bir **liste** olarak yazılır. Listeler H6'da ele alınır.

| Excel (Türkçe) | Excel (İngilizce) | Python |
|---|---|---|
| `=TOPLA(C9:C11)` | `=SUM(C9:C11)` | `sum(faizler)` |
| `=ORTALAMA(C9:C11)` | `=AVERAGE(C9:C11)` | `sum(faizler) / len(faizler)` |
| `=MİN(C9:C11)` | `=MIN(C9:C11)` | `min(faizler)` |
| `=MAK(C9:C11)` | `=MAX(C9:C11)` | `max(faizler)` |

`len(faizler)` listedeki eleman sayısıdır (burada 3).

**Tahmin sorusu:** Toplam kaçtır? Anapara ile toplam faiz toplanınca ne çıkar?

```python
# Excel Hesap!B9:B11: her yılın bileşik bakiyesi (döngüsüz, Excel'deki gibi satır satır)
y1 = anapara * (1 + oran)     # Excel B9:  =B8*(1+$B$2)
y2 = y1 * (1 + oran)          # Excel B10: =B9*(1+$B$2)
y3 = y2 * (1 + oran)          # Excel B11: =B10*(1+$B$2)

# Excel Hesap!C9:C11: her yılın bileşik faizi (=B9-B8 kalıbı)
faizler = [y1 - anapara, y2 - y1, y3 - y2]
print(faizler)
print(sum(faizler))                   # Excel: =TOPLA(C9:C11)     İngilizce: =SUM(C9:C11)
print(sum(faizler) / len(faizler))    # Excel: =ORTALAMA(C9:C11)  İngilizce: =AVERAGE(C9:C11)
print(min(faizler))                   # Excel: =MİN(C9:C11)       İngilizce: =MIN(C9:C11)
print(max(faizler))                   # Excel: =MAK(C9:C11)       İngilizce: =MAX(C9:C11)

# [S] Sağlama: Excel Hesap!C12 ve Hesap!B38 (=B1+C12-B11)
excel_toplam_faiz = 17440             # Excel Hesap!C12
assert abs(sum(faizler) - excel_toplam_faiz) < 0.01
assert abs(anapara + sum(faizler) - bilesik) < 0.01
print("Python'daki toplam faiz Excel'deki C12 ile aynı; anapara + toplam faiz = son bakiye.")
```

# Faiz türleri ve teknik hesaplar

Bu bölüm Excel'deki **Faiz** sayfasının çalışılmış örneklerini Python'da yeniden kurar. Her adımda aynı düzen vardır: **[G]** girdiler, **[H]** hesap, **[S]** Excel'deki değerle sağlama (`assert`).

**Hızlı şerit:** Python bilenler bu bölümü hızlıca çalıştırıp doğrudan **Uygulama G3-G4** bölümüne geçebilir.

| Python | Excel (Faiz sayfası) | Anlamı |
|---|---|---|
| `P` | B31 | anapara (100.000 TL) |
| `r` | B32 | yıllık nominal oran (0.40) |
| `n` | B33 | süre (3 yıl) |

Yeni araçlar:
- `import math`: matematik modülünü yükler. `math.exp(x)` Excel'deki ÜS (EXP), yani e^x; `math.log(x)` Excel'deki LN, yani doğal logaritmadır.
- `f"{x:,.2f}"`: sayıyı binlik ayırıcı ve 2 ondalıkla yazar. Python İngilizce biçim kullanır: 274,400.00. `f"{x:.2%}"` oranı yüzde olarak yazar: 0.4821 için 48.21%.

```python
# [G] Girdiler: Excel Faiz!B31:B33
import math

P = 100000   # anapara, TL          (Excel Faiz!B31)
r = 0.40     # yıllık nominal oran  (Excel Faiz!B32)
n = 3        # süre, yıl            (Excel Faiz!B33)
print(P, r, n)
```

## Isınma: iki finansal okuryazarlık sorusu (Faiz sayfası, 1. bölüm)

Lusardi ve Mitchell'in finansal okuryazarlık sorularından ("Big Three") ikisinin uyarlaması. Kaynak: GFLEC, https://gflec.org/wp-content/uploads/2015/04/3-Questions-Article2.pdf

1. Bir hesapta 100 TL bulunmakta, faiz yıllık %2'dir. Para 5 yıl hiç çekilmezse hesapta 102 TL'den fazla mı, tam 102 TL mi, az mı olur?
2. Hesabın faizi yıllık %1, enflasyon yıllık %2'dir. Bir yıl sonra bu parayla bugünkünden fazla mı, aynı mı, az mı şey alınabilir?

**Tahmin sorusu:** İki sorunun cevabı nedir? Cevaplar not edilir, sonra hücre çalıştırılır.

```python
# Excel Faiz!B22 =B19*(1+B20)^B21   ve   Faiz!B27 =(1+B25)/(1+B26)-1
soru1 = 100 * (1 + 0.02) ** 5          # 100 TL, yıllık %2, 5 yıl
soru2 = (1 + 0.01) / (1 + 0.02) - 1    # faiz %1, enflasyon %2: satın alma gücündeki değişim
print(f"Soru 1: {soru1:.2f} TL")
print(f"Soru 2: {soru2:.2%}")

# [S] Sağlama: Excel Faiz!B22 ve B27
assert abs(soru1 - 110.41) < 0.01
assert abs(soru2 - (-0.0098)) < 0.0001
print("Sağlama tamam: 1. soru 102 TL'den fazla, 2. soruda satın alma gücü azalır.")
```

## [H] Basit, bileşik ve sürekli bileşik faiz (Faiz sayfası, 2. bölüm)

Aynı "yıllık %40" üç farklı biçimde işleyebilir:
- **Basit:** faiz yalnız anaparaya işler. P × (1 + r × n)
- **Bileşik:** her yılın faizi anaparaya eklenir. P × (1 + r)^n
- **Sürekli bileşik:** faizin her an eklendiği varsayılır; bileşiklenme sıklığının sınırıdır. P × e^(r × n)

**Tahmin sorusu:** Sürekli bileşik, yıllık bileşikten çok mu fazla, az mı?

```python
# [H] Excel Faiz!B38 =B31*(1+B32*B33)   B39 =B31*(1+B32)^B33   B41 =B31*EXP(B32*B33)
basit = P * (1 + r * n)
bilesik = P * (1 + r) ** n
surekli_tutar = P * math.exp(r * n)       # Excel: ÜS (EXP)
print(f"Basit          : {basit:,.2f} TL")
print(f"Yıllık bileşik : {bilesik:,.2f} TL")
print(f"Sürekli bileşik: {surekli_tutar:,.2f} TL")

# [S] Sağlama: Excel Faiz!B38, B39, B41
assert abs(basit - 220000) < 0.01
assert abs(bilesik - 274400) < 0.01
assert abs(surekli_tutar - 332011.69) < 0.01
print("Sağlama tamam: Excel Faiz!B38, B39, B41 ile aynı.")
```

## [H] Bileşiklenme sıklığı (Faiz sayfası, 3. bölüm)

Yıllık nominal %40, yılda m kez eklenirse dönem oranı r / m, yıllık etkin oran (1 + r / m)^m - 1 olur. Excel'de formül satır satır aşağı sürüklenmiştir. Python'da aynı işi `for` döngüsü yapar: köşeli parantezdeki her `m` için gövde bir kez çalışır.

`etkinler = {}` bir **sözlük**tür: her `m` için bulunan oranı `etkinler[m]` adıyla saklar. Sözlükler H6'da ayrıntılı ele alınır.

**Tahmin sorusu:** Aylık eklemede (m = 12) yıllık etkin oran yüzde kaç? Sürekli bileşik bundan kaç puan fazla?

```python
# [H] Excel Faiz!C50:F53   C =$B$32/B50   D =(1+C50)^B50-1   F =$B$31*(1+D50)
print("m        dönem oranı   yıllık etkin   1 yıl sonunda (TL)")
etkinler = {}
for m in [1, 2, 4, 12]:
    donem_orani = r / m
    etkin_oran = (1 + donem_orani) ** m - 1
    etkinler[m] = etkin_oran
    print(f"{m:<8} {donem_orani:>11.4%} {etkin_oran:>14.4%} {P * (1 + etkin_oran):>20,.2f}")

etkin_surekli = math.exp(r) - 1           # Excel Faiz!D54 =EXP($B$32)-1
print(f"{'sürekli':<8} {'':>11} {etkin_surekli:>14.4%} {P * math.exp(r):>20,.2f}")

# [S] Sağlama: Excel Faiz!D53 (aylık) ve D54 (sürekli)
assert abs(etkinler[12] - 0.482126) < 0.000001
assert abs(etkin_surekli - 0.491825) < 0.000001
print("Sağlama tamam: Excel Faiz!D53 ve D54 ile aynı.")
```

## [H] Nominal ve etkin: ETKİN ve NOMİNAL'in Python karşılığı (Faiz sayfası, 4. bölüm)

numpy-financial kütüphanesinde ETKİN (EFFECT) ve NOMİNAL (NOMINAL) yoktur; formülle yazılır:
- etkin = (1 + nominal / m)^m - 1
- nominal = m × ((1 + etkin)^(1/m) - 1)

`** (1 / m)` m'inci dereceden kök almaktır. Excel'deki ETKİN ve NOMİNAL m'yi tam sayıya keser; bu formüller kesmez (bu fark G2'de görülür).

```python
# [H] Excel Faiz!B64 =B61*B62   B65 =(1+B61)^B62-1   B66 =EFFECT(B64,B62)   B67 =NOMINAL(B65,B62)
aylik = 0.03          # Excel Faiz!B61
m = 12                # Excel Faiz!B62
nominal = aylik * m
etkin_oran = (1 + nominal / m) ** m - 1                # ETKİN (EFFECT) formülü
nominal_geri = m * ((1 + etkin_oran) ** (1 / m) - 1)   # NOMİNAL (NOMINAL) formülü
print(f"Yıllık nominal       : {nominal:.4%}")
print(f"Yıllık etkin         : {etkin_oran:.4%}")
print(f"Etkinden geri nominal: {nominal_geri:.4%}")

# [S] Sağlama: Excel Faiz!B64, B65, B67
assert abs(nominal - 0.36) < 0.000001
assert abs(etkin_oran - 0.425761) < 0.000001
assert abs(nominal_geri - nominal) < 0.000001
print("Sağlama tamam: Excel Faiz!B64, B65, B67 ile aynı.")
```

## [H] Reel faiz (Faiz sayfası, 5. bölüm)

**Fisher eşitliği:** (1 + nominal) = (1 + reel) × (1 + enflasyon). Buradan reel = (1 + nominal) / (1 + enflasyon) - 1. **Kısa yol:** nominal - enflasyon; düşük oranlarda Fisher'e yakındır, yüksek oranlarda yanıltır.

Kaynaklar:
- Politika faizi (bir hafta vadeli repo ihale faiz oranı) %37: TCMB PPK kararı, 10.09.2026. https://www.tcmb.gov.tr/wps/wcm/connect/tr/tcmb+tr/main+menu/duyurular/basin/2026/duy2026-38
- Yıllık TÜFE, Eylül 2026 %29,73 ve Ağustos 2026 %31,51: TÜİK verisi, TCMB Tüketici Fiyatları tablosu, erişim 06.10.2026. https://www.tcmb.gov.tr/wps/wcm/connect/TR/TCMB+TR/Main+Menu/Istatistikler/Enflasyon+Verileri/Tuketici+Fiyatlari

**Tahmin sorusu:** Kısa yol mu büyük çıkar, Fisher mi?

```python
# [G] Kaynaklar yukarıdaki metin hücresinde
politika = 0.37        # Excel Faiz!B74
tufe_eylul = 0.2973    # Excel Faiz!B75
tufe_agustos = 0.3151  # Excel Faiz!B76

# [H] Excel Faiz!B77 =(1+B74)/(1+B75)-1   B78 =(B74-B75)*B34   B80 =(1+B74)/(1+B76)-1
reel_eylul = (1 + politika) / (1 + tufe_eylul) - 1
kisa_yol = politika - tufe_eylul
reel_agustos = (1 + politika) / (1 + tufe_agustos) - 1
print(f"Reel faiz, Fisher (Eylül)  : {reel_eylul:.2%}")
print(f"Kısa yol (Eylül)           : {kisa_yol * 100:.2f} puan")   # 100: oran farkını puana çevirme (Excel B34)
print(f"Reel faiz, Fisher (Ağustos): {reel_agustos:.2%}")

# [S] Sağlama: Excel Faiz!B77, B78, B80
assert abs(reel_eylul - 0.056039) < 0.000001
assert abs(kisa_yol * 100 - 7.27) < 0.000001
assert abs(reel_agustos - 0.041746) < 0.000001
print("Sağlama tamam: Excel Faiz!B77, B78, B80 ile aynı.")
```

## [H] İkiye katlanma süresi (Faiz sayfası, 6. bölüm)

(1 + r)^T = 2 denkleminin iki yanının doğal logaritması alınır: T × ln(1 + r) = ln 2, yani T = ln 2 / ln(1 + r). Excel'de süre TAKSİT_SAYISI (NPER) ya da LN ile bulunmuştur. **72 kuralı** pratik bir yaklaşımdır: T ≈ 72 / (yüzde olarak oran).

**Tahmin sorusu:** %40'ta 72 kuralı tam süreden uzun mu, kısa mı?

```python
# [H] Excel Faiz!B90 =LN(B86)/LN(1+B32)   B91 =B87/(B32*B34)
T_tam = math.log(2) / math.log(1 + r)
T_72 = 72 / (r * 100)                 # 72 kuralı oranı yüzde sayısı olarak ister (40)
print(f"Tam süre : {T_tam:.4f} yıl")
print(f"72 kuralı: {T_72:.4f} yıl")
print(f"Hata     : {T_72 - T_tam:.4f} yıl")

# [S] Sağlama: Excel Faiz!B89 (TAKSİT_SAYISI) ve B91; (1 + r)^T gerçekten 2 mi?
assert abs(T_tam - 2.0600) < 0.0001
assert abs(T_72 - 1.80) < 0.000001
assert abs((1 + r) ** T_tam - 2) < 1e-12
print("Sağlama tamam: Excel Faiz!B89 ve B91 ile aynı.")
```
