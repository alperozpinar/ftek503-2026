## [G] Girdiler

Excel'de girdiler Hesap sayfasının B1, B2 ve B3 hücrelerindeydi. Python'da her girdiye bir **ad** veririz. Bu adlara **değişken** denir.

| Excel | Python | Değer |
|---|---|---|
| B1 (anapara) | `anapara` | 10000 |
| B2 (yıllık oran, %40) | `oran` | 0.40 |
| B3 (süre, yıl) | `yil` | 3 |

Okurken dikkat edin:
- `anapara = 10000` satırındaki `=` "eşittir" değil, "sağdaki değeri soldaki ada ver" demektir. Buna **atama** denir. Excel'de formülün başındaki `=` ile karıştırmayın.
- `#` işaretinden sonrası **açıklamadır**. Python o kısmı çalıştırmaz; insanlar için yazılır.
- Python'da ondalık ayırıcı **noktadır**: `0.40`. Türkçe Excel'deki `0,40` Python'da `0.40` yazılır.
- %40, Python'da `0.40` yazılır. Excel'de de hücrede 0,40 saklanıyordu; yüzde yalnız bir görünüştü.

```python
# [G] Girdiler: Excel'deki Hesap sayfasının B1, B2, B3 hücreleriyle aynı değerler
anapara = 10000   # TL          (Excel: B1)
oran = 0.40       # yıllık %40  (Excel: B2)
yil = 3           # yıl         (Excel: B3)
print(anapara, oran, yil)
```

## [H] Basit faiz

Basit faizde faiz yalnız anaparaya işler. Excel'de B5'e `=B1*(1+B2*B3)` yazmıştık ve sonuç 22.000 TL idi.

**Tahmin et:** Aşağıdaki hücre ne yazdıracak? Tahmininizi not edin, sonra çalıştırın.

```python
# [H] Basit faiz. Excel (Türkçe ve İngilizce aynı): =B1*(1+B2*B3)
basit = anapara * (1 + oran * yil)
print(basit)
```

## [H] Bileşik faiz

Bileşik faizde her yılın faizi anaparaya eklenir ve ertesi yıl o da faiz kazanır. Excel'de B6'ya `=B1*(1+B2)^B3` yazmıştık ve sonuç 27.440 TL idi.

Python'da üs işareti `^` değil, **iki yıldız** `**`.

**Tahmin et:** Aşağıdaki hücre tam olarak `27440` mı yazdıracak?

```python
# [H] Bileşik faiz. Excel: =B1*(1+B2)^B3 ; Python'da üs işareti ** (iki yıldız)
bilesik = anapara * (1 + oran) ** yil
print(bilesik)
```

### İncele: neden 27439.999999999993?

- Bilgisayar ondalık sayıları ikilik sistemde saklar. 0.40 gibi bazı sayılar ikilik sistemde tam yazılamaz; çok küçük bir yuvarlama payıyla saklanır. Bunu 1/3'ün ondalık olarak 0,3333... diye bitmeden gitmesine benzetebilirsiniz.
- Bu küçük pay çarpmalar boyunca taşınır. Sonuç 27.440'tan yaklaşık 0,000000000007 kadar sapar. Para hesabında önemsizdir.
- Excel'de hücre iki ondalıkla biçimlendirildiği için orada 27.440,00 görürsünüz.
- Bu yüzden Excel ile Python'u karşılaştırırken "tam eşit mi?" diye değil, **"fark çok küçük mü?"** diye sorarız. Ayrıntısı H6'da.
- Python'da `^` üs demek değildir, başka bir işlemdir. `(1 + oran) ^ yil` yazarsanız Python bir hata mesajı (`TypeError`) verir. Üs için her zaman `**` kullanın.

## [S] Sağlama: Python ile Excel aynı sonucu veriyor mu?

Üç yeni araç:
- `abs(x)`: x'in işaretini atar. Excel'deki MUTLAK (ABS) fonksiyonu gibidir. `abs(-3)` sonucu 3'tür.
- `==` (çift eşittir): iki tarafın eşit olup olmadığını **sorar**. Cevap `True` (doğru) ya da `False` (yanlış) olur; Excel'deki mantıksal değer gibi (Hesap!B53). Tek `=` atamadır, çift `==` karşılaştırmadır.
- `assert koşul`: "bu koşul doğru olmalı" demektir. Koşul doğruysa hiçbir şey olmaz. Yanlışsa Python `AssertionError` hatasıyla durur.

Aşağıda fark 0,01'den (1 kuruştan) küçükse sağlama geçer.

**Tahmin et:** İlk `print` satırı `True` mu `False` mu yazdıracak?

```python
# [S] Sağlama: Excel'in verdiği sonuçlarla karşılaştırma
excel_basit = 22000     # Excel Hesap!B5
excel_bilesik = 27440   # Excel Hesap!B6

print(bilesik == excel_bilesik)          # tam eşit mi?
print(abs(bilesik - excel_bilesik))      # aradaki fark ne kadar?
assert abs(basit - excel_basit) < 0.01       # fark 1 kuruştan küçükse sessizce geçer
assert abs(bilesik - excel_bilesik) < 0.01
print("Sağlama tamam: Python ile Excel aynı sonucu veriyor.")
```

**İncele:** İkinci satırdaki `7.275957614183426e-12` bilimsel gösterimdir: 7,28 × 10⁻¹², yani yaklaşık 0,00000000000728. Sıfır değildir ama 1 kuruştan çok çok küçüktür. Bu yüzden `==` `False` verdi, `assert` ise geçti.

## [H] Yıl yıl tablo

Excel'de B9'a `=B8*(1+$B$2)` yazıp aşağı sürüklemiştiniz. `$B$2` sayesinde oran her satırda aynı hücreden geliyordu. Python'da "aynı işi her yıl için tekrarla" demenin yolu `for` döngüsüdür.

**`for` döngüsünü H6'da öğreneceğiz.** Şimdilik hücreyi yalnız çalıştırın ve çıktıyı Excel'deki tabloyla (Hesap!A8:F11) karşılaştırın.

`round(x, 2)`: x'i 2 ondalığa yuvarlar, Excel'deki YUVARLA (ROUND) gibi. İkisi arasındaki küçük farkı H6'da göreceğiz.

**Tahmin et:** Son satırdaki bileşik bakiye kaç çıkacak? Fark hangi yıldan itibaren sıfırdan büyük olacak?

```python
# [H] Yıl yıl tablo. Excel: B9 =B8*(1+$B$2)   D9 =$B$1*(1+$B$2*A9)   F9 =B9-D9
# for döngüsünü H6'da öğreneceğiz; şimdilik yalnız çalıştırın.
bakiye = anapara                          # Excel B8 =B1 (0. yıl)
print("Yıl", "Bileşik", "Basit", "Fark")
print(0, bakiye, anapara, 0)
for t in range(1, yil + 1):               # t sırayla 1, 2, 3 olur: Excel'de bir satır aşağı inmek gibi
    bakiye = bakiye * (1 + oran)          # oran her satırda aynı: Excel'deki $B$2
    basit_t = anapara * (1 + oran * t)    # basit faiz, t. yılın sonunda
    print(t, round(bakiye, 2), round(basit_t, 2), round(bakiye - basit_t, 2))
```

```python
# [S] Sağlama: tablonun son satırı tek hücre formülüne eşit mi? (Excel Hesap!B36: =B11-B6)
print(bakiye - bilesik)
assert abs(bakiye - bilesik) < 0.01
print("Tablonun son satırı tek hücre formülüyle aynı.")
```

## [H] TOPLA, ORTALAMA, MİN, MAK'ın Python karşılığı

Excel'de her yılın bileşik faizi Hesap!C9:C11 hücrelerindeydi: 4.000, 5.600 ve 7.840 TL. Aşağıda önce her yılın bakiyesi Excel'deki gibi bir öncekinden hesaplanıyor (B9, B10, B11). Sonra her yılın faizi bulunup (C9:C11) köşeli parantez içinde bir **liste** olarak yazılıyor. Listeleri H6'da öğreneceğiz.

| Excel (Türkçe) | Excel (İngilizce) | Python |
|---|---|---|
| `=TOPLA(C9:C11)` | `=SUM(C9:C11)` | `sum(faizler)` |
| `=ORTALAMA(C9:C11)` | `=AVERAGE(C9:C11)` | `sum(faizler) / len(faizler)` |
| `=MİN(C9:C11)` | `=MIN(C9:C11)` | `min(faizler)` |
| `=MAK(C9:C11)` | `=MAX(C9:C11)` | `max(faizler)` |

`len(faizler)` listedeki eleman sayısıdır (burada 3).

**Tahmin et:** Toplam kaç? Anapara ile toplam faiz toplanınca ne çıkar?

```python
# Excel Hesap!B9:B11: her yılın bileşik bakiyesi (döngüsüz, Excel'deki gibi satır satır)
y1 = anapara * (1 + oran)     # Excel B9:  =B8*(1+$B$2)
y2 = y1 * (1 + oran)          # Excel B10: =B9*(1+$B$2)
y3 = y2 * (1 + oran)          # Excel B11: =B10*(1+$B$2)

# Excel Hesap!C9:C11: her yılın bileşik faizi (=B9-B8 kalıbı)
faizler = [y1 - anapara, y2 - y1, y3 - y2]
print(faizler)
print(sum(faizler))                   # Excel: =TOPLA(C9:C11)     İngilizce: =SUM(C9:C11)
print(sum(faizler) / len(faizler))    # Excel: =ORTALAMA(C9:C11)  İngilizce: =AVERAGE(C9:C11)
print(min(faizler))                   # Excel: =MİN(C9:C11)       İngilizce: =MIN(C9:C11)
print(max(faizler))                   # Excel: =MAK(C9:C11)       İngilizce: =MAX(C9:C11)

# [S] Sağlama: Excel Hesap!C12 ve Hesap!B38 (=B1+C12-B11)
excel_toplam_faiz = 17440             # Excel Hesap!C12
assert abs(sum(faizler) - excel_toplam_faiz) < 0.01
assert abs(anapara + sum(faizler) - bilesik) < 0.01
print("Python'daki toplam faiz Excel'deki C12 ile aynı; anapara + toplam faiz = son bakiye.")
```

## Değiştir

Şimdi sayıları siz değiştireceksiniz. Aşağıdaki hücre yukarıdaki hesapları bozmasın diye yeni adlar kullanıyor (`yil_yeni`, `oran_yeni`).

**Değiştir 1:** `yil_yeni = 3` satırını `yil_yeni = 5` yapın ve hücreyi çalıştırın. Sonra Excel'de Hesap sayfasında B3'ü 5 yapın; B5 ve B6 ile karşılaştırın. Excel'deki [S] Sağlama bloğunda ne oldu? Neden? İşiniz bitince Excel'de B3'ü 3'e geri alın.

**Değiştir 2:** `yil_yeni` değerini yeniden 3 yapın ve `oran_yeni = 0.40` satırını `oran_yeni = 0.30` (yüzde 30) yapın. Fark (faizin faizi) büyüdü mü küçüldü mü? Tahmin edin, sonra çalıştırın.

**Değiştir 3:** İkisini birlikte değiştirin: 5 yıl ve yüzde 30.

Excel'deki sonucu Python'la karşılaştırmayı unutmayın: iki taraf aynı girdilerle aynı sonucu vermeli.

```python
# DEĞİŞTİR: yalnız bu iki satırdaki sayıları değiştirin, sonra hücreyi çalıştırın.
yil_yeni = 3        # Değiştir 1: 5 yapın
oran_yeni = 0.40    # Değiştir 2: 0.30 yapın

basit_yeni = anapara * (1 + oran_yeni * yil_yeni)        # Excel: =B1*(1+B2*B3)
bilesik_yeni = anapara * (1 + oran_yeni) ** yil_yeni     # Excel: =B1*(1+B2)^B3
print("Basit  :", round(basit_yeni, 2))
print("Bileşik:", round(bilesik_yeni, 2))
print("Fark   :", round(bilesik_yeni - basit_yeni, 2))
```
